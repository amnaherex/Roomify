## Numerical vs Target analysis

```python
df.groupby('room_type')['price'].agg(['mean', 'median', 'count'])
```

```python
df.groupby('room_type')['minimum_nights'].agg(['mean', 'median', 'count'])
```

relationship appears weaker than the relationship between price and room type

```python
df.groupby('room_type')['number_of_reviews'].agg([
    'mean',
    'median',
    'count'
])

sns.boxplot(
    data=df,
    x='room_type',
    y='number_of_reviews',
    showfliers=False
)

plt.title('Number of Reviews vs Room Type')
plt.show()
```

```python
df.groupby('room_type')['reviews_per_month'].agg([

    'mean',

    'median',

    'count'

]) 

sns.boxplot(

    data=df,

    x='room_type',

    y='reviews_per_month',

    showfliers=False

)



plt.title('Reviews per Month vs Room Type')

plt.show()
```

```python
df.groupby('room_type')['calculated_host_listings_count'].agg([
    'mean',
    'median',
    'count'
])
sns.boxplot(
    data=df,
    x='room_type',
    y='calculated_host_listings_count',
    showfliers=False
)

plt.title('Host Listings Count vs Room Type')
plt.show()
```

```python
df.groupby('room_type')['availability_365'].agg([
    'mean',
    'median',
    'count'
])
sns.boxplot(
    data=df,
    x='room_type',
    y='availability_365',
    showfliers=False
)

plt.title('Availability vs Room Type')
plt.show()
```

```python
correlation_matrix = df[numeric_cols].corr()

correlation_matrix
```

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Numerical Feature Correlation Matrix')
plt.show()
```

```python
df['last_review'] = pd.to_datetime( df['last_review'],errors='coerce')
```

```python
df['last_review'].describe()
```

* Number of reviews vs room type: Private rooms have slightly more reviews on average, while Shared rooms have the fewest, so number_of_reviews 
shows only a weak difference across room types.
* Reviews per month vs room type: Shared and Private rooms receive slightly more reviews per month than Entire homes, but the difference is not 
very large.
* Calculated host listings count vs room type: Entire home/apt listings are associated with hosts having a much higher average number of listings, 
although the median is still low, indicating some hosts with very large portfolios.
* Availability 365 vs room type: Shared rooms are available for noticeably more days per year, while Entire homes and Private rooms have very similar
availability.
* Numerical correlation matrix: Most numerical features have weak correlations with each other; the strongest relationship is between 
number_of_reviews and reviews_per_month at about 0.55.
* Last review: last_review contains 10,052 missing values, and the available review dates range from 2011-03-28 to 2019-07-08.
